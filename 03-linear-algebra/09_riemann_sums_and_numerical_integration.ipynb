```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
```

```python
def f(x):
    return x ** 2

def left_riemann(f, a, b, n):
    dx = (b - a) / n
    xs = np.linspace(a, b, n, endpoint=False)

    return np.sum(f(xs)) * dx


for n in [4, 10, 50]:
    print(f"n = {n},  {left_riemann(f, 0, 3, n)}")
```

```python
# task 1 same logic with different values

def f(x):
    return x

def left_riemann(f,a, b, n):
    dx = (b - a) / n
    xs = np.linspace (a, b, n, endpoint=False)

    return np.sum(f(xs)) * dx


for n in [4, 8, 100]:
    print(f"n = {n},  {left_riemann(f, 0, 3, n)}")
```

```python
# graph of the task 1
a = 0
b = 3
n = 8

# for left riemann graph
dx = (b - a) / n
xs = np.linspace(a, b, n)

# for actual graph
x = np.linspace(a, b, 100)

plt.plot(x, f(x), label="f(x) = x")

plt.bar(xs, f(xs), width=dx, edgecolor="black", align="edge")

plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Left Riemann")
plt.legend()
plt.show()
```
