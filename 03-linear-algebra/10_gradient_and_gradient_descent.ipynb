## Task 1 — Kafe foydasi

$P(x, y) = 60x + 50y - 2x^2 - y^2 - xy$

```python
# task 1 kafe foydasi
# P(x, y) = 60x + 50y − 2x2 − y2 − xy

# a) qolda deyilgan , bu commentda hisoblash keremi yoki limitni formulasi blanmi?
# dP_dx = 60 - 4x - y => x boyicha(yani y constanta)
# dP_dy = 50 - 2y - x => y boyicha(yani x constanta)


def P(x, y):
    return 60 * x + 50 * y - 2 * x ** 2 - y ** 2 - x * y

h = 0.00001

def gradient (f, x, y, h):
    df_dx = (f(x + h, y) - f(x, y)) / h
    df_dy = (f(x , y + h) - f(x, y)) / h

    return np.array([df_dx, df_dy])
```

**b)** Hozirgi holatda (x=5, y=10) gradientni hisoblash

```python
# b) Hozir kafe x = 5, y = 10 sotmoqda

print(gradient(P, 5, 10, h))

# x boyicha qisman derivative kattaro chiqdi, shunga x ni oshirsak koproq tasir qiladi modelga
```

**c)** Foyda maksimal bo'ladigan (x, y) nuqtani topish (sympy)

```python
# c) foyda eng katta bo'ladigan (x, y)

# agar qolda hisoblasa
# dP_dx = 60 - 4x - y
# dP_dx = 50 - 2y - x
# shulani nolga tenglab sistema ishlash kere => x = 10 va y = 20 shunda P ni topsak 800 bolarkan

x, y = sp.symbols("x y")
P = 60 * x + 50 * y - 2 * x ** 2 - y ** 2 - x * y

df_dx = sp.diff(P, x)
df_dy = sp.diff(P, y)

natija = sp.solve(
    [df_dx, df_dy],
    [x, y]
)
print(natija) # {x: 10, y: 20}

profit = P.subs(natija) # subs funksiya blan x va y solamiz
print(profit)
```

## Task 3 — Imtihon ballini bashorat qilish

$ball = w \cdot soat + b$, $\quad L(w, b) = (w \cdot 3 + b - 75)^2$

**a)** Model bashorati va loss

**b)** $\partial L/\partial w$ va $\partial L/\partial b$ ni qo'lda hisoblash

```python
# task 3 Imtihon ballini bashorat qilish

# ball = w·soat + b.
# L(w, b) = (w·3 + b − 75) ** 2

soat = 3
ball = 75
w = 10
b = 20

def L(w, b):
    return (w * 3 + b - 75) ** 2

# a) Model qancha ball bashorat qiladi? Loss qancha?
ball = w * soat + b
loss = L(w, b)
print(ball, " ", loss) # 50   625
# ball = 50, loss = 625

# b) ∂L/∂w va ∂L/∂b ni qolda hisoblang

# dL_dw = 6 * (w·3 + b − 75) => w boyicha
# dL_db = 2 * (w·3 + b − 75) => b boyicha
```

**c)** lr = 0.01 bilan bitta gradient descent qadami

**d)** Nega $\partial L/\partial w$ moduli $\partial L/\partial b$ nikidan 3 marta katta?

```python
# c) lr = 0.01 bilan bitta gradient descent qadamini bajaring. Yangi w, b va yangi loss ni toping.

lr = 0.01

def gradient(w, b):
    dL_dw = np.mean(2 * soat * (w * 3 + b - 75))
    dL_db = np.mean(2 * (w * 3 + b - 75))

    return np.array([dL_dw, dL_db])

dL_dw , dL_db = gradient(w, b)

# w blan b ni bitta qadamdan keyingi valuelari
new_w = w - lr * dL_dw
new_b = b - lr * dL_db

print(new_w, " ", new_b) # 11.5   20.5


# d) Nega ∂L/∂w ning moduli ∂L/∂b nikidan 3 marta katta?
# chunki formulada
# dL_dw = np.mean(2 * soat * (w * 3 + b - 75))
# dL_db = np.mean(2 * (w * 3 + b - 75))
# chunki dL_dw da soatga kopaytiralgan, bizda esa soat 3 soat
```

## Task 4 — Toshkent kvartira narxlari (mini-loyiha)

$narx = w \cdot maydon + b$

**a)-f)** Gradient descent turli learning rate'lar bilan, natijani `np.polyfit` bilan solishtirish va 65 m² uchun bashorat

```python
# task 4 Toshkent kvartira narxlari (mini-loyiha)
# narx = w·maydon + b

maydon = np.array([4, 5, 6, 7, 8])
narxi = np.array([30, 36, 44, 50, 58])

def gradient(w, b):
    df_dw = np.mean(2 * maydon * (w * maydon + b - narxi))
    df_db = np.mean(2 * (w * maydon + b - narxi))

    return np.array([df_dw, df_db])

# a)
# b) 10-, 100-, 1000- va 5000-qadamlarda w, b va cost ni chop qiling
qadam = 5000
lr = [0.005, 0.02, 0.03]


for x in lr:
    w = 0
    b = 0
    for step in range(1, qadam + 1):
        df_dw, df_db = gradient(w, b)


        # w va b ni yangilash
        w = w - x * df_dw
        b = b - x * df_db

        # cost topish
        prediction = w * maydon + b
        cost = np.mean((prediction - narxi) ** 2)


        # b) da soralgan narsa
        if step == 10 or step == 100 or step == 1000 or step == 5000:
            print(f"lr = {x} step = {step} w = {w}, b = {b}, cost = {cost}")



# c)  lr = 0.03 da nima bo'ldi? Nega?
# step = 10 w = -122.00177611863066, b = -19.274516682077177, cost = 372129.0600436089
# step = 100 w = -28860454096383.297, b = -4563079925963.89, cost = 1.8604228222243473e+28
# step = 1000 w = -9.013531696581643e+126, b = -1.4251149829228203e+126, cost = 1.8146602195810028e+255
# step = 5000 w = nan, b = nan, cost = nan

# qadam kattalik qildi, portlash yuzaga keldi



# d) w tez o'rnashadi, b esa sekin. Nima uchun?
# chunki formulada w da yana maydonga kopaytiriladi shunga
# w va b ning yaqinlashish tezligi feature scale va gradientlarga bog'liq.


# e) Natijani np.polyfit(maydon, narx, 1) bilan solishtirin
print(np.polyfit(maydon, narxi, 1)) # [7.  1.6]


# f) 65 m² kvartira narxini bashorat qiling.
w = 7
b = 1.6

maydon_65 = 6.5
prediction = w * maydon_65 + b

print(prediction) # 47.1 ming dollar
```

## Task 5 — Ikki xususiyatli uy modeli

$narx = w_1 \cdot maydon + w_2 \cdot xonalar + b$

**a)-d)** Cost funksiyasi hosilalari, gradient descent, natijalarning talqini va gradient vektor o'lchami

```python
# task 5 Ikki xususiyatli uy modeli
# narx = w1·maydon + w2·xonalar + b

maydon = np.array([4, 6, 5, 8])
xonalar = np.array([2, 3, 2, 4])
narx = np.array([36, 49, 41, 62])


def gradient(w1, w2, b):
    prediction = w1 * maydon + w2 * xonalar + b
    error = prediction - narx

    df_dw1 = np.mean(2 * maydon * error)
    df_dw2 = np.mean(2 * xonalar * error)
    df_db = np.mean(2 * error)

    return np.array([df_dw1, df_dw2, df_db])


# a)
# C(w1, w2, b) = mean((prediction - narx)^2)
#
# dC/dw1 = mean(2 * maydon * error)
# dC/dw2 = mean(2 * xonalar * error)
# dC/db = mean(2 * error)


# b) Gradient descent
qadam = 5000
lr = 0.02

w1 = 0
w2 = 0
b = 0

for step in range(1, qadam + 1):

    # gradient topish
    df_dw1, df_dw2, df_db = gradient(w1, w2, b)

    # cost topish
    prediction = w1 * maydon + w2 * xonalar + b
    cost = np.mean((prediction - narx) ** 2)

    # w1, w2 va b ni yangilash
    w1 = w1 - lr * df_dw1
    w2 = w2 - lr * df_dw2
    b = b - lr * df_db

    # b) da soralgan qadamlar
    if step == 100 or step == 1000 or step == 5000:
        print(f"step = {step}, w1 = {w1}, w2 = {w2}, b = {b}, cost = {cost}")


# c) Yakuniy qiymatlar real hayotda nimani anglatadi?
# w1 => maydon 1 birlikka oshsa, narx qancha ozgaradi
# w2 => xonalar 1 taga oshsa, narx qancha ozgaradi
# b => maydon va xonalar 0 bo'lgandagi boshlang'ich qiymat


# d) Gradient vektor necha o'lchamli?
# 3 o'lchamli: [df_dw1, df_dw2, df_db]
# Agar yana 2 ta xususiyat qo'shsak:
# 5 o'lchamli gradient bo'ladi
```
